```python
import numpy as np
from qutip import qeye, sigmax, entropy_vn

I, sx = qeye(2), sigmax()

def rho_q_analytic(a):
    c = float(np.exp(-2*abs(a)**2))    
    return 0.5*(I + c*sx)

a = np.linspace(0, 5, 100)
S = np.array([entropy_vn(rho_q_analytic(x)) for x in a])
```

```python
plt.figure(figsize=(10, 6))
plt.plot(a, S, label='von Neumann Entropy', color='blue')
plt.title('von Neumann Entropy of the Qubit State as a Function of a')
plt.xlabel('a')
plt.ylabel('Entropy S')
plt.grid()
plt.legend()    
```
